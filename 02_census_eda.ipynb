```python
import pandas as pd, geopandas as gpd

census = pd.read_csv("data/san-diego-census-2017/San Diego.csv", dtype={"GEOID": str})
print("Columns:", census.columns.tolist())

# geometry is stored as WKT text in one column, so build the GeoDataFrame from it
geom_col = [c for c in census.columns
            if census[c].astype(str).str.startswith(("POLYGON", "MULTIPOLYGON")).any()][0]
gdf = gpd.GeoDataFrame(census,
                       geometry=gpd.GeoSeries.from_wkt(census[geom_col]),
                       crs="EPSG:3857").to_crs("EPSG:4326")
gdf["GEOID"] = gdf["GEOID"].str.zfill(11)   # restore any dropped leading 0 (CA tracts start with 06)
print("GEOID sample:", gdf["GEOID"].head(3).tolist())
```

```python
import pandas as pd, urllib.request, ssl, certifi, io

url = "https://lehd.ces.census.gov/data/lodes/LODES7/ca/wac/ca_wac_S000_JT00_2019.csv.gz"

# macOS Python often ships without a CA bundle, so hand urllib certifi's
ctx = ssl.create_default_context(cafile=certifi.where())
with urllib.request.urlopen(url, context=ctx) as resp:
    data = resp.read()

wac = pd.read_csv(io.BytesIO(data), compression="gzip", dtype={"w_geocode": str})

sd = wac[wac["w_geocode"].str.startswith("06073")].copy()   # San Diego County
sd["GEOID"] = sd["w_geocode"].str[:11]                       # block -> tract
jobs = (sd.groupby("GEOID", as_index=False)["C000"]
          .sum().rename(columns={"C000": "total_jobs"}))
print(jobs.shape[0], "tracts with jobs")
```

```python
import pandas as pd
meta = pd.read_csv("data/san-diego-census-2017/metaindicator.csv")
print(meta)                       # maps each code/column -> what it measures
print("\nColumns:", gdf.columns.tolist())
```

```python
# --- resolve the population column whether your data uses the code or a short name ---
pop_var = "B01003_001E"                       # ACS total population
row = meta[meta["var_id"] == pop_var].iloc[0]
candidates = [pop_var] + [str(v) for v in row.values if isinstance(v, str)]
POP_COL = next((c for c in candidates if c in gdf.columns), None)
print("Population column:", POP_COL)          # should print a real name, not None

# --- merge jobs onto census tracts ---
merged = gdf.merge(jobs, on="GEOID", how="left")
merged["total_jobs"] = merged["total_jobs"].fillna(0)
print("joined:", merged["total_jobs"].gt(0).sum(), "/", len(merged), "tracts")

# --- blended demand score (weights become Alice's sliders) ---
def norm(s):
    return (s - s.min()) / (s.max() - s.min())

merged["pop_norm"]  = norm(merged[POP_COL].astype(float))
merged["jobs_norm"] = norm(merged["total_jobs"])

w_pop, w_jobs = 0.5, 0.5
merged["demand_score"] = w_pop * merged["pop_norm"] + w_jobs * merged["jobs_norm"]
print(merged[["GEOID", POP_COL, "total_jobs", "demand_score"]].head())
```

```python
import pandas as pd, geopandas as gpd, os

census = pd.read_csv("data/san-diego-census-2017/San Diego.csv", dtype={"GEOID": str})

# WKT geometry is Web Mercator (EPSG:3857); declare that, then convert to lat/lon
geom_col = [c for c in census.columns
            if census[c].astype(str).str.startswith(("POLYGON", "MULTIPOLYGON")).any()][0]
gdf = gpd.GeoDataFrame(census,
                       geometry=gpd.GeoSeries.from_wkt(census[geom_col]),
                       crs="EPSG:3857").to_crs("EPSG:4326")
gdf["GEOID"] = gdf["GEOID"].str.zfill(11)

os.makedirs("data/processed", exist_ok=True)
gdf.to_parquet("data/processed/census_clean.parquet")
print("saved", len(gdf), "tracts")
```
